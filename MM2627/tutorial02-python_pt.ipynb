## Execução Condicional: ```if```, ```elif``` and ```else```
O Python possui duas formas de execução condicional: a mais comum usa blocos de código iniciados por linhas com as *keywords* `if`, `elif` e `else`. O conteúdo do bloco de código abaixo de uma linha `if` ou `elif` só é executado se a condição estabelecida nessa linha for verdadeira. A instrução `else` permite criar um bloco de código que é executado quando todas as condições ditadas nos blocos `if` e `elif` precedentes são falsas (último recurso).

````python
a = 1 
if a < 5:
    print("a é menor que 5")
````

Um **bloco de código** é um conjunto de inhas de código que se encontra com um nível de indentação diferente das anteriores. No exemplo acima, os quatro espaços que precedem a instrução `print` são significativos e indicam que esta linha de códifo está dependente da instrução `if`.

````python
a = 1 
if a > 5:
    print("a é maior que 5") # this line won't be exectuded
else:
    print("a não é maior do que 5") 
````

As instruções `elif` são colocadas após um bloco `if` para indicar uma ou mais alternativas a correr mediante a veracidade de outra(s) condição(ões). Podemos usar vários blocos `elif` em sucessão, mas **apenas o bloco dependente da primeira condição verdadeira é executado**.

````python
a = 1
if a < 5:
  print("a é menor que 5")
elif a < 2: 
  print("a é menor que 2") 
elif a > 10:
  print("a é maior que 10")
else:
  print("a está entre 5 e 10, inclusive")
````

## Blocos de Código
Um aspeto peculiar do Python é o facto de o código ser dividido em blocos, e cada bloco ser identificado pela sua *indentação* (ou seja, o número de carateres de espaço entre o início da linha e o primeiro carater que não seja um espaço nessa mesma linha). No exemplo anterior, cada linha com uma instrução `print()` é um bloco, associado ao cabeçalho do bloco anterior (a instrução `if`, `elif` ou `else`). Em geral, os blocos de código são precedidos por dois pontos (:) no final do seu respetivo cabeçalho. Assim, uma forma abstrata de denotar as estruturas `if/elif/else` é através do seguinte pseudocódigo:

```
if (condição A):
   bloco a executar se a condição A for verdadeira
elif (condição B):
   bloco a executar se a condição B for verdadeira
elif (condição C):
   bloco a executar se a condição C for verdadeira
...
else:
   bloco a executar se nenhuma das condições acima for verdadeira

```

O Python permite a utilização de qualquer número de carateres de espaço ou tabulações para a indentação, desde que seja mantido o mesmo padrão ao longo do código. Neste *notebook*, a indentação utilizada é de dois (2) carateres de espaço. O exemplo de pseudocódigo acima utiliza 3 espaços. Também é comum utilizarem-se 4 espaços ou 1 carater de tabulação. A norma do Python PEP 008 contém um conjunto de instruções e boas práticas no que diz respeito à indentação de código.

## Execução Repetitiva: ciclos `for` e `while`
O Python oferece duas formas de execução repetitiva:
1. Ciclos `for`, normalmente usados quando sabemos quantas vezes o código vai ser repetido (por exemplo, quando um dado código deve atuar em cada um dos elementos de uma lista), ou quando pretendemos impor um limite rígido ao número de repetições a executar.
2. Ciclos `while`, normalmente usados quando desejamos repetir um bloco de código até que uma dada condição seja verificada, mas não sabemos *a priori* quantas repetições serão necessárias.

### Ciclos `for`
A instrução `for` percorre todos os valores de um objeto que tenha a propriedade de ser iterável. Genericamente, é usado da seuinte forma:
```
for iter_var in iterable:
   bloco a executar
```
em cada repetiçaõ a variável `iter_var` toma valores sucessivos do iterável. Exemplos de objetos iteráveis são:
* Listas
* Dicionários (itera ao longo da coleçao de chaves/*keys*)
* Texto (itera um carater de cada vez)

Exemplo:

````python
lst_objetos = ['livro', 'estojo', 'carro'] # lista de objectos

for objecto in lst_objetos:
  print("O objecto agora é um", objecto)
````

````python
for c in 'hello':
    print(c.upper())
````

### Ciclos `while`
Os ciclos `while` são usados para repetir um bloco de código até que uma dada condição seja verdadeira. A sua sintaxe genérica é:
```
while (condição):
    bloco de código a executar
```


````python
# Exemplo: determinar o primeiro número inteiro cujo inverso é menor que 0.025
n=1
while 1/n >= 0.025:
  n = n+1 # vamos incrementando o valor de n
print(n,1/n)
````

No exemplo acima, note que a linha com a instrução `print()` não está indentada, indicando que só será executada **depois** do ciclo `while` terminar.

O exemplo acima pode ser escrito usando um ciclo `for` com um número pre-determinado de iterações grande e uma instrução `break` para travar a execução assim que a condição for verificada, conforme pode verificar abaixo. No entanto, esta abordagem torna o código mais confuso e difícil de entender.

````python
for n in range(1,1000):
  if 1/n < 0.025:
    break
print(n, 1/n)
````

## Reciclagem de código: Funções
No python, uma função é um bloco de código com um nome definido e que pode ser invocado várias vezes após a sua definição. Cada função deve ter um nome próprio e ser definida num bloco `def`. Por exemplo:

````python
def euDigoOla():
  print("Olá!\nEu sou a funçao que diz olá!")
````

A função `euDigoOla()` pode agora ser executada (invocada) sempre que necessário.

````python
euDigoOla()
````

As funções podem receber dados do exterior através dos seus argumentos (valores ou variáveis que colocamos entre os parentísis curvos). Por exemplo, a função abaixo é decladada para operar num argumento (`n`) o qual não é definido no momento da criação da função.

````python
def imprimeQuadrado(n):
  # n não é definido, mas definimos como a função vai processar n
  print(f"O quadrado de {n} é {n**2}")
````

````python
imprimeQuadrado(2) # we call the function, and n take the value of 2
````

O uso de argumentos permite à função receber dados do mundo exterior. Mas as funções definidas acima apenas imprimem um resultado no ecrã. Quando desejamos que a função comunique o resultado do seu trabalho com o mundo exterior, devemos usar a instrução `return` com o valor a devolver. A instrução `return` termina instantaneamente a execução da função.

O exemplo abaixo calcula a média dos valores de uma lista. Como pode ver nos exemplos a seguir, o resultado da função pode depois ser impresso ou sofrer novas operações.

````python
def media(lista):
  "Devolve a média dos valores de uma lista"
  n = len(lista)
  return sum(lista)/n
````

````python
myList = [2,3,6,9,8,4,12,3]
print("média de myList: ", media(myList) )
print("quadrado da média: ", media(myList)**2 )
````

Podemos devolver vários valores aomundo exterior, se os colocarmos numa lista, num dicionário ou numa tupla (uma forma simplificada de listas). Note que uma função pode invocar outras funções, permitindo criar operações complexas a partir de blocos de operações mais simples.

````python
def minMedMax(lista):
  "Devolve o mínimo, média e máximo de uma lista"
  # we can call our functions from wihtin other functions
  return (min(lista), media(lista), max(lista))
````

````python
minMedMax([2,3,6,9,8,4,12,3])
````

O elemento de texto "pendurado" na primeira linha do bloco da função é a chamada ```docString```. Este é o texto que a função  `help()` usa:

````python
help(media)
````

````python
help(minMedMax)
````

Finalmente, uma função pode invocar-se a si mesma. Este comportamento chama-se **recursão** e, embora seja permitido e por vezes seja a única forma de resolver um problema, o seu uso carrega um custo elevado.

````python
def factorial(n):
  if (n == 1):
    return 1
  else:
    return n*factorial(n-1)
````

A função factorial definida acima de forma recursiva encontra-se definida na forma mais pareccida com a sua definição aritmética, no entanto, ela também pode ser definida numa forma não recursiva:

````python
def factorialNR(n):
    o=1
    for i in range(1,n+1):
        o = o*i
    return o
````

#### Álgebra Linear com o Numpy
Podemos usar objetos do tipo `array` para representar vectores. O Numpy possui funcões dedicadas ao cálculo da norma e do produto escalar. A partir daqui, calcular o ângulo entre dois vectores é trivial.

````python
# distância entre dois pontos no espaço (i.e. norma da diferença entre dois vectores)
v1 = np.array([8.2,6.1,3.0])
v2 = np.array([1.0, 0.0, 0.0])
np.linalg.norm(v1)
````

````python
# produto escalar entre vetores
np.dot(v1,v2)
````

````python
# angulo entre vectores
amp = np.arccos(np.dot(v1,v2)/(np.linalg.norm(v1)*np.linalg.norm(v2)))
print(f"A amplitude é: {np.rad2deg(amp)}")
````

#### Números Aleatórios
Gerar números aleatórios é vital em várias aplicações da modelação molecular.
No entanto, se usarmos números totalmente aleatórios, nenhuma simulação será reproduzivel, já que o gerador de números aleatórios não vai gerar duas vezes a mesma sequência de valores aleatórios.

````python
# using the random sub-module of Numpy
x = np.random.random()
print("Um número aleatório entre 0 e 1",x)
x = np.random.random(3)
print("Uma array com 3 números aleatórios entre 0 e 1",x)
x = np.random.normal()
print("Um número aleatório extraído de uma distribuição normal de média 0 e desvio padrão 1",x)
x = np.random.normal(5,0.5,10)
print("Uma array com 10 números aleatórios extraídos de uma distribuição normal de média 5 e desvio padrão 0.5",x)
````

Sempre que corremos a célula acima vamos encontrar números diferentes. Isto traz uma desvantagem em Modelação Molecular: a falta de reprodutibilidade!

Para obtermos resultados reproduzíveis devemos ter uma fonte (gerador) de números aleatórios previsível (ou seja, os números devem ser aleatórios, mas ser gerados numa sequência repetível). Para isso, criamos um gerador de números aleatórios (com a função ``numpy.random.default_rng()`` fornecendo-lhe um número inteiro que sirva de semente). O objecto gerador de núeros aleatórios tem os mesmos métodos para gerar números aleatórios que o sub-módulo ``numpy.random``.

````python
gerador = np.random.default_rng(42)
x = gerador.random()
print("Um número aleatório entre 0 e 1",x)
x = gerador.random(3)
print("Uma array com 3 números aleatórios entre 0 e 1",x)
x = gerador.normal()
print("Um número aleatório extraído de uma distribuição normal de média 0 e desvio padrão 1",x)
x = gerador.normal(5,0.5,10)
print("Uma array com 10 números aleatórios extraídos de uma distribuição normal de média 5 e desvio padrão 0.5",x)
````

Agora sempre que corrermos a célula acima, os números aleatórios serão sempre iguais!

### O Módulo Scikit-learn
O pacote `scikit-learn` contém uma coleção curada de algoritmos usados em *Machine Learning*, os quais são implementados usando uma interface comums. Isto permite aplicar diversos métodos ao mesmo problema fazendo apenas pequenas mudanças no código.

As células abaixo exemplificam a aplicação de um procedimento de regressão linear aos dados sobre pinguins importados na secção do seaborn.

````python
penguins.head() # inspect the forst few lines of the data set
````

````python
p = penguins.dropna() # remove rows with missing values
````

````python
from sklearn import linear_model # import the sub-module with the model we whant to use

#define the model as an object
reg = linear_model.LinearRegression()

#fit the model, by providing the independent variables and the dependent variable
reg.fit(p[['flipper_length_mm','body_mass_g']],
        p['bill_length_mm']) 
````

````python
# we can use the score() method to have an idea of how well the model behaves
reg.score(p[['flipper_length_mm','body_mass_g']],p['bill_length_mm'])
````

````python
# finally we do a dispersion plot of the model's predictions (y) and the actual data (x)

plt.clf()
x = np.linspace(p['bill_length_mm'].min(),p['bill_length_mm'].max(),100) # this is to trace an identity line


preds = reg.predict(p[['flipper_length_mm','body_mass_g']]) # the model's predictions

fig, ax = plt.subplots()
ax.scatter(p['bill_length_mm'],preds, label="Linear Model Prediction") # the dispersion plot
ax.plot(x,x) # an idendity line for guidance
ax.set_xlabel("Observed") # set labels for the axes
ax.set_ylabel("Predicted")
ax.legend() # place a key on the plot
plt.show()
````

O modelo de regressão linear não é muito bom, como acontece frequentemente nestes estudos.

O exemplo (bastante complexo) abaixo cria um modelo mais complexo que tem em conta alguns dados de texto (como a espécie dos pinguins e a ilha onde vivem), bem como um tratamento polinomial das variáveis numéricas. Isto requer a implementação de filtros para as colunas e a aplicação de um tratamento diferente a cada coluna dependendo do seu conteúdo. Utilizamos o algoritmo AdaBoost (adaptive boosting), que é muito mais flexível do que uma regressão linear. Como será discutido mais à frente no curso, os algoritmos de machine learning flexíveis têm tendência a sobreajustar-se aos dados nos quais são treinados, pelo que reservamos 30% dos dados originais para testar o modelo após o ajuste e avaliar o quão bem o modelo consegue fazer previsões com base em dados com os quais não foi treinado.

````python
from sklearn import pipeline
from sklearn import preprocessing
from sklearn import compose
from sklearn import model_selection
from sklearn import ensemble
tratador_numeros = pipeline.Pipeline(steps=
[('normalizar', preprocessing.StandardScaler()),
('polinomial', preprocessing.PolynomialFeatures(degree=2))])

tratador_texto = preprocessing.OneHotEncoder(drop='if_binary',
handle_unknown='ignore')

filtro = compose.ColumnTransformer([
('cats', tratador_texto,
compose.make_column_selector(dtype_include=object)),
('nums', tratador_numeros,
compose.make_column_selector(dtype_include=np.number))],
remainder='passthrough')

modelo = pipeline.Pipeline(steps=[
('Pre-tratamento', filtro),
('Regressão', ensemble.AdaBoostRegressor() )])

# divisão dos dados ente treino e teste:
treino, teste = model_selection.train_test_split(p,test_size=0.30)

features=['species', 'island', 'bill_depth_mm', 'flipper_length_mm',
'body_mass_g', 'sex'] # variaveis explicativas
target=['bill_length_mm'] # variável-alvo

X_train = treino[features]
y_train = treino[target]

modelo.fit(X_train,y_train.to_numpy().ravel())
````

````python
modelo.score(X_train,y_train)
````

````python
# agora para o conjunto de teste...
X_test = teste[features]
y_test = teste[target]
modelo.score(X_test,y_test)
````

````python
plt.clf()
previsoes_treino = modelo.predict(X_train)
previsoes_teste = modelo.predict(X_test)
fig, ax = plt.subplots()
ax.scatter(y_train,previsoes_treino,c='blue',label='Train data')
ax.scatter(y_test,previsoes_teste,c='red',label='Test data')
ax.plot(x,x) # fazer uma linha de identidade
ax.set_xlabel("Observed")
ax.set_ylabel("Predicted")
ax.legend()
plt.show()
````

Como podemos ver, as previsões do modelo AdaBoost encontram-se mais concentradas junto da linha de identidade, apesar de se manteres alguns casos extremos.
