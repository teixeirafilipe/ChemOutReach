## Conditional Execution: `if`, `elif` and `else`
Python has two forms of conditional execution: the most common uses code blocks started by lines with the keywords `if`, `elif`, and `else`. The content of the code block below an `if` or `elif` line is only executed if the condition established in that line is true. The `else` statement allows creating a code block that is executed when all conditions dictated in the preceding `if` and `elif` blocks are false (last resort).

````python
a = 1 
if a < 5:
    print("a is less than 5")
````

A **code block** is a set of code lines that are at a different indentation level than the preceding ones. In the example above, the four spaces preceding the `print` statement are significant and indicate that this line of code depends on the `if` statement.

````python
a = 1 
if a > 5:
    print("a is greater than 5") # this line won't be executed
else:
    print("a is not greater than 5")
````

The `elif` statements are placed after an `if` block to indicate one or more alternatives to run depending on the truth value of other condition(s). We can use multiple `elif` blocks in succession, but **only the block dependent on the first true condition is executed**.

````python
a = 1
if a < 5:
  print("a is less than 5")
elif a < 2: 
  print("a is less than 2") 
elif a > 10:
  print("a is greater than 10")
else:
  print("a is between 5 and 10, inclusive")
````

## Code Blocks
One peculiar aspect of Python is the fact that code is divided into blocks, and each block is identified by its *indentation* (i.e. the number of space characters between the start of the line and the first non-space character in the line). In the previous example, each line with a `print()` statement is a block, associated with its previous block header (the `if`, `elif` or `else` statement). In general, code blocks are preceded by a colon (:) at the end of their corresponding header. Thus, an abstract way to denote the `if/elif/else` constructs is by using the following pseudo-code:

```
if (condition A):
   block to execute if condition A is true
elif (condition B):
   block to execute if condition B is true
elif (condition C):
   block to execute if condition C is true
...
else:
   block to execute if none of the conditions above are true
```

Python will allow any number of space or tab characters to be used for indentation, as long as it is the same throughout the code. In this *notebook*, the indentation used is two (2) space characters. The pseudo-code example above uses 3 spaces. It is also common to have 4 spaces or 1 tab character. Python standard PEP 008 contains a number of instructions and best practices with respect to code indentation.

## Repetitive Execution: `for` and `while` loops
Python offers two forms of repetitive execution:
1. `for` loops, typically used when we know how many times the code will be repeated (for example, when a given code should act on each element of a list), or when we want to impose a strict limit on the number of repetitions to execute.
2. `while` loops, typically used when we wish to repeat a code block until a given condition is met, but we do not know *a priori* how many repetitions will be necessary.

### `for` Loops
The `for` statement iterates through all values of an object that has the property of being iterable. Generally, it is used in the following way:
```
for iter_var in iterable:
   block to execute
```
in each repetition the variable `iter_var` takes successive values from the iterable. Examples of iterable objects are:
* Lists
* Dictionaries (iterates over the collection of keys)
* Text (iterates one character at a time)

Example:

````python
lst_objects = ['book', 'pencil case', 'car'] # list of objects

for object in lst_objects:
  print("The object now is a", object)
````

````python
for c in 'hello':
    print(c.upper())
````

### `while` Loops
`while` loops are used to repeat a code block until a given condition is true. Its generic syntax is:
```
while (condition):
    code block to execute
```


````python
# Example: determine the first integer whose inverse is less than 0.025
n=1
while 1/n >= 0.025:
  n = n+1 # we increment the value of n
print(n,1/n)
````

In the example above, note that the line with the `print()` statement is not indented, indicating that it will only be executed **after** the `while` loop finishes.

The example above can be written using a `for` loop with a large predetermined number of iterations and a `break` statement to stop execution as soon as the condition is verified, as you can check below. However, this approach makes the code more confusing and harder to understand.

````python
for n in range(1,1000):
  if 1/n < 0.025:
    break
print(n, 1/n)
````

In Python, a function is a code block with a defined name that can be invoked multiple times after its definition. Each function must have its own name and be defined inside a `def` block. For example:

The use of arguments allows the function to receive data from the outside world. But the functions defined above only print a result on the screen. When we want the function to communicate the result of its work to the outside world, we must use the `return` statement with the value to return. The `return` statement instantly terminates the execution of the function.

The example below calculates the average of values in a list. As you can see in the following examples, the result of the function can then be printed or undergo new operations.

````python
def mean(list):
  "Returns the average of the values in a list"
  n = len(list)
  return sum(list)/n
````

````python
myList = [2,3,6,9,8,4,12,3]
print("mean of myList: ", mean(myList) )
print("square of mean: ", mean(myList)**2 )
````

We can return multiple values to the outside world if we put them in a list, dictionary, or tuple (a simplified form of lists). Note that a function can invoke other functions, allowing the creation of complex operations from simpler blocks of operations.

````python
def minMeanMax(list):
  "Returns the minimum, mean, and maximum of a list"
  # we can call our functions from within other functions
  return (min(list), mean(list), max(list))
````

````python
minMeanMax([2,3,6,9,8,4,12,3])
````

Finally, a function can invoke itself. This behavior is called **recursion** and, although permitted and sometimes the only way to solve a problem, its use carries a high cost.

````python
def factorial(n):
  if (n == 1):
    return 1
  else:
    return n*factorial(n-1)
````

The factorial function defined recursively above is structured in the form closest to its arithmetic definition; however, it can also be defined in a non-recursive form:

````python
def factorialNR(n):
    o=1
    for i in range(1,n+1):
        o = o*i
    return o
````

#### Linear Algebra with Numpy
We can use `array` objects to represent vectors. Numpy has functions dedicated to computing norms and dot products. From there, calculating the angle between two vectors is trivial.

````python
# distance between two points in space (i.e. norm of the difference between two vectors)
v1 = np.array([8.2,6.1,3.0])
v2 = np.array([1.0, 0.0, 0.0])
np.linalg.norm(v1)
````

````python
# dot product between vectors
np.dot(v1,v2)
````

````python
# angle between vectors
amp = np.arccos(np.dot(v1,v2)/(np.linalg.norm(v1)*np.linalg.norm(v2)))
print(f"The amplitude is: {np.rad2deg(amp)}")
````

#### Random Numbers
Generating random numbers is vital in various molecular modeling applications.
However, if we use completely random numbers, no simulation will be reproducible, since the random number generator will not generate the same sequence of random values twice.

````python
# using the random sub-module of Numpy
x = np.random.random()
print("A random number between 0 and 1",x)
x = np.random.random(3)
print("An array with 3 random numbers between 0 and 1",x)
x = np.random.normal()
print("A random number drawn from a normal distribution with mean 0 and standard deviation 1",x)
x = np.random.normal(5,0.5,10)
print("An array with 10 random numbers drawn from a normal distribution with mean 5 and standard deviation 0.5",x)
````

Every time we run the cell above, we will get different numbers. This brings a disadvantage in Molecular Modeling: the lack of reproducibility!

To obtain reproducible results, we must have a predictable source (generator) of random numbers (that is, the numbers must be random, but generated in a repeatable sequence). To do this, we create a random number generator (with the function `numpy.random.default_rng()` providing an integer as a seed). The random number generator object has the same methods for generating random numbers as the `numpy.random` sub-module.

````python
generator = np.random.default_rng(42)
x = generator.random()
print("A random number between 0 and 1",x)
x = generator.random(3)
print("An array with 3 random numbers between 0 and 1",x)
x = generator.normal()
print("A random number drawn from a normal distribution with mean 0 and standard deviation 1",x)
x = generator.normal(5,0.5,10)
print("An array with 10 random numbers drawn from a normal distribution with mean 5 and standard deviation 0.5",x)
````

Now every time we run the cell above, the random numbers will always be the same!

### The Scikit-learn Module
The `scikit-learn` package contains a curated collection of algorithms used in *Machine Learning*, which are implemented using a common interface. This allows applying various methods to the same problem by making only minor changes to the code.

The cells below exemplify the application of a linear regression procedure to the penguin data imported in the seaborn section.

````python
penguins.head() # inspect the first few lines of the data set
````

````python
p = penguins.dropna() # remove rows with missing values
````

````python
from sklearn import linear_model # import the sub-module with the model we want to use

#define the model as an object
reg = linear_model.LinearRegression()

#fit the model, by providing the independent variables and the dependent variable
reg.fit(p[['flipper_length_mm','body_mass_g']],
        p['bill_length_mm'])
````

````python
# we can use the score() method to have an idea of how well the model behaves
reg.score(p[['flipper_length_mm','body_mass_g']],p['bill_length_mm'])
````

````python
# finally we do a dispersion plot of the model's predictions (y) and the actual data (x)

plt.clf()
x = np.linspace(p['bill_length_mm'].min(),p['bill_length_mm'].max(),100) # this is to trace an identity line


preds = reg.predict(p[['flipper_length_mm','body_mass_g']]) # the model's predictions

fig, ax = plt.subplots()
ax.scatter(p['bill_length_mm'],preds, label="Linear Model Prediction") # the dispersion plot
ax.plot(x,x) # an identity line for guidance
ax.set_xlabel("Observed") # set labels for the axes
ax.set_ylabel("Predicted")
ax.legend() # place a key on the plot
plt.show()
````

The linear regression model is not very good, as often happens in these studies.

The (rather complex) example below creates a more complex model that takes into account some text data (such as the penguin species and the island where they live), as well as a polynomial treatment of numerical variables. This requires implementing column filters and applying a different treatment to each column depending on its content. We use the AdaBoost (adaptive boosting) algorithm, which is much more flexible than a linear regression. As will be discussed later in the course, flexible machine learning algorithms tend to overfit the data they are trained on, so we reserve 30% of the original data to test the model after fitting and evaluate how well the model can make predictions based on data it was not trained on.

````python
from sklearn import pipeline
from sklearn import preprocessing
from sklearn import compose
from sklearn import model_selection
from sklearn import ensemble
number_handler = pipeline.Pipeline(steps=
[('normalize', preprocessing.StandardScaler()),
('polynomial', preprocessing.PolynomialFeatures(degree=2))])

text_handler = preprocessing.OneHotEncoder(drop='if_binary',
handle_unknown='ignore')

filter_col = compose.ColumnTransformer([
('cats', text_handler,
compose.make_column_selector(dtype_include=object)),
('nums', number_handler,
compose.make_column_selector(dtype_include=np.number))],
remainder='passthrough')

model = pipeline.Pipeline(steps=[
('Pre-treatment', filter_col),
('Regression', ensemble.AdaBoostRegressor() )])

# splitting the data between train and test:
train, test = model_selection.train_test_split(p,test_size=0.30)

features=['species', 'island', 'bill_depth_mm', 'flipper_length_mm',
'body_mass_g', 'sex'] # explanatory variables
target=['bill_length_mm'] # target variable

X_train = train[features]
y_train = train[target]

model.fit(X_train,y_train.to_numpy().ravel())
````

````python
model.score(X_train,y_train)
````

````python
# now for the test set...
X_test = test[features]
y_test = test[target]
model.score(X_test,y_test)
````

````python
plt.clf()
train_predictions = model.predict(X_train)
test_predictions = model.predict(X_test)
fig, ax = plt.subplots()
ax.scatter(y_train,train_predictions,c='blue',label='Train data')
ax.scatter(y_test,test_predictions,c='red',label='Test data')
ax.plot(x,x) # make an identity line
ax.set_xlabel("Observed")
ax.set_ylabel("Predicted")
ax.legend()
plt.show()
````

As we can see, the predictions of the AdaBoost model are more concentrated along the identity line, although some extreme cases remain.
